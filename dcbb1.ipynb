```python
import numpy as np
import matplotlib.pyplot as plt

# Time vector
t = np.arange(0, 1, 0.0001)

# User inputs
Am = float(input("Message amplitude (Am): "))
Ac = float(input("Carrier amplitude (Ac): "))
Fm = float(input("Message frequency (Fm): "))
Fc = float(input("Carrier frequency (Fc): "))

# Message signal
Mt = Am * np.cos(2 * np.pi * Fm * t)

# Carrier signal
Ct = Ac * np.cos(2 * np.pi * Fc * t)

# Modulation Index
Ma = Am / Ac

# Amplitude Sensitivity
Ka = Ma / Am

# AM Signal
S = Ac * (1 + Ka * Mt) * np.cos(2 * np.pi * Fc * t)

# Plotting
plt.figure(figsize=(10, 8))
```

```python

# Message Signal
plt.subplot(3, 1, 1)
plt.plot(t, Mt, linewidth=2)
plt.title("Message Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

# Carrier Signal
plt.subplot(3, 1, 2)
plt.plot(t, Ct, 'r', linewidth=2)
plt.title("Carrier Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

# AM Signal
plt.subplot(3, 1, 3)
plt.plot(t, S, 'g', linewidth=2)
plt.title("Amplitude Modulated (AM) Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()

# Display Modulation Index
print("Modulation Index (Ma):", Ma)
```
